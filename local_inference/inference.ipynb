# Local Bridge Model Inference

This notebook only loads the Qwen base model and LoRA adapter. If `./base_model/` is incomplete, the first load downloads the base model there; later runs reuse the local files.

Place the adapter files in `./adapter/` before running the Python cells. The adapter is never downloaded automatically.

```python
from pathlib import Path

# Edit paths here only. Paths are relative to this notebook's folder.
BASE_MODEL_ID = "Qwen/Qwen2.5-1.5B-Instruct"
BASE_MODEL_PATH = Path("./base_model")
ADAPTER_PATH = Path("./adapter")
DOWNLOAD_BASE_MODEL = True
LOAD_IN_4BIT = True
MAX_NEW_TOKENS = 400

print("Base model:", BASE_MODEL_PATH.resolve())
print("Adapter:", ADAPTER_PATH.resolve())
print("Download missing base model:", DOWNLOAD_BASE_MODEL)
```

```python
from model_loader import ModelConfig, generate_text, load_model_and_tokenizer, select_device

config = ModelConfig(
    base_model_id=BASE_MODEL_ID,
    base_model_path=BASE_MODEL_PATH,
    adapter_path=ADAPTER_PATH,
    download_base_model=DOWNLOAD_BASE_MODEL,
    load_in_4bit=LOAD_IN_4BIT,
    max_input_tokens=2048,
)

print("Device:", select_device())
model, tokenizer = load_model_and_tokenizer(config)
print("Base model and LoRA adapter loaded.")
```

```python
def ask(prompt, max_new_tokens=MAX_NEW_TOKENS, temperature=0.1, top_p=0.9):
    return generate_text(
        model,
        tokenizer,
        prompt,
        max_new_tokens=max_new_tokens,
        temperature=temperature,
        top_p=top_p,
    )
```

```python
examples = [
    "What is Lacey's formula for normal scour depth D below the foundation design discharge level?",
    "What are the grip length rules for open foundations in ordinary soil, hard rock, and soft rock?",
    "Generate JSON structural parameters for an RCC single-cell box culvert. Use only the keys needed by a CAD constraint engine.",
]

for prompt in examples:
    print("PROMPT:", prompt)
    print(ask(prompt))
    print("-" * 80)
```
